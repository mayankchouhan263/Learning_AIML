```python
from bs4 import BeautifulSoup

with open("scraped_data/data1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
all_h3 = soup.find_all("h3")
all_countries = []
for h3 in all_h3:
    name = h3.get_text(strip=True)
    population = (h3.find_next("div").select_one("span.country-population").get_text(strip=True))
    all_countries.append([name,population])
```

```python
import pandas as pd 
df = pd.DataFrame(all_countries,columns= ["Name","Population"])
```

```python
df.to_csv("Cleaned_data/data.csv",index= False )
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
page_count = 1
while True :
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)
    
    soup = BeautifulSoup(res.text,"lxml")
    quotes = soup.select("div.quote")
    if not quotes:
        print(f"total pages = {page_count - 1} ")
        break 
    with open(f"scraped_data/quotes on page{page_count}.html", "wb") as f:
        f.write(res.content)
    print(f"Data dounloaded from page {page_count}")
    page_count += 1
```

```python
#Extract Information 
```

```python
page_count = 1
quotes = []
while page_count<11:
    with open(f"scraped_data/quotes_on_page{page_count}.html", "rb") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")
    for aq in all_quotes:
        #auther = (aq.find("small", class_="author").text)
        author = (aq.select_one("small.author").get_text())
        quote = (aq.select_one("span.text").get_text())
        tags = [tag.get_text(strip=True) for tag in aq.select("a.tag")]
        quotes.append([author,quote,tags])
    page_count +=1
```

```python
import pandas as pd 
df = pd.DataFrame(quotes,columns= ["auther","Quotes","tags"])
```
