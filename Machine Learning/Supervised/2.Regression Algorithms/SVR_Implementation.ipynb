```python
from sklearn.svm import SVR 
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler
from sklearn import datasets 
```

```python
df = datasets.load_diabetes(as_frame= True ).frame
```

```python
df.head()
```

```python
X = df.drop("target" , axis = 1)
y = df["target"]
```

```python
X_train , X_test , y_train , y_test = train_test_split(
    X,y,  test_size=0.30, random_state=42
)
```

```python
y_scaler = StandardScaler()

y_train_scaled = y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled = y_scaler.transform(y_test.values.reshape(-1,1)).ravel()
```

```python
model = SVR()

model.fit(X_train,y_train_scaled)
```

```python
y_prediction_scaled = model.predict(X_test)
y_train_prediction_scaled = model.predict(X_train)

print("r2 score ",r2_score(y_test_scaled,y_prediction_scaled))
print("r2 score ",r2_score(y_train_scaled,y_train_prediction_scaled))
```

```python
#Linear Kernel 
model = SVR(kernel = "linear")

model.fit(X_train,y_train_scaled)
y_prediction_scaled = model.predict(X_test)
y_train_prediction_scaled = model.predict(X_train)

print("r2 score ",r2_score(y_test_scaled,y_prediction_scaled))
print("r2 score ",r2_score(y_train_scaled,y_train_prediction_scaled))
```

```python
#Linear Kernel 
model = SVR(kernel = "poly")

model.fit(X_train,y_train_scaled)
y_prediction_scaled = model.predict(X_test)
y_train_prediction_scaled = model.predict(X_train)

print("r2 score ",r2_score(y_test_scaled,y_prediction_scaled))
print("r2 score ",r2_score(y_train_scaled,y_train_prediction_scaled))
```

# HYPERPARAMETER TUNIG GRID SEARCH CV

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "C": [1, 2, 5, 10, 50, 100],
    "kernel": ["rbf","linear" , "poly"],
    "epsilon": [0.01,0.1,0.2,0.3,0.5] 
}
```

```python
svr = SVR()

grid_search = GridSearchCV(svr,param_grid,scoring = "r2" , cv = 5)

grid_search.fit(X_train , y_train_scaled)
```

```python
print("BEST PARAMETERS : ",grid_search.best_params_)
```

```python
#BEST MODEL 

best_model = SVR(C = 10, epsilon = 0.1, kernel = 'linear')

best_model.fit(X_train,y_train_scaled)
y_prediction_scaled = best_model.predict(X_test)
y_train_prediction_scaled = best_model.predict(X_train)

print("r2 score for testing data :",r2_score(y_test_scaled,y_prediction_scaled))
print("r2 score for training data :",r2_score(y_train_scaled,y_train_prediction_scaled))
```

# LINEAR SVR

```python
from sklearn.svm import LinearSVR

lsvr_model = LinearSVR(C = 10, epsilon = 0.1 , max_iter= 5000)

lsvr_model.fit(X_train,y_train_scaled)
y_prediction_scaled = lsvr_model.predict(X_test)
y_train_prediction_scaled = lsvr_model.predict(X_train)

print("r2 score for testing data :",r2_score(y_test_scaled,y_prediction_scaled))
print("r2 score for training data :",r2_score(y_train_scaled,y_train_prediction_scaled))
```
