```python
import numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
```

```python
df = datasets.load_iris(as_frame = True).frame
```

```python
df.head()
```

```python
X = df.drop("target" , axis = 1)
y = df["target"]
```

```python
X_train,X_test ,y_train ,y_test  = train_test_split(
    X , y ,test_size = 0.3 ,random_state = 42 , stratify = y
)
```

```python
#SCAILING THE DATA 

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_trained_scale = scaler.fit_transform(X_train)

X_test_scale = scaler.transform(X_test) 
```

```python
# Build SVC 
svc = SVC()
svc.fit(X_trained_scale , y_train )
```

```python
y_pred = svc.predict(X_test_scale)
```

```python
#EVALUATE 

from sklearn.metrics import accuracy_score , classification_report  ,f1_score

print ("ACCURACY = ",accuracy_score(y_test , y_pred))
print ("classification_report = ",classification_report(y_test , y_pred))
print ("f1_score = ",f1_score(y_test , y_pred ,  average='weighted'))
```

```python
# LINEAR KERNAL 

svc = SVC(kernel = "linear")
svc.fit(X_trained_scale , y_train )
y_pred = svc.predict(X_test_scale)

print ("ACCURACY = ",accuracy_score(y_test , y_pred))
```

```python
# LINEAR KERNAL 

svc = SVC(kernel = "poly")
svc.fit(X_trained_scale , y_train )
y_pred = svc.predict(X_test_scale)

print ("ACCURACY = ",accuracy_score(y_test , y_pred))
```

```python
# LINEAR KERNAL 

svc = SVC(kernel = "sigmoid")
svc.fit(X_trained_scale , y_train )
y_pred = svc.predict(X_test_scale)

print ("ACCURACY = ",accuracy_score(y_test , y_pred))
```

```python
C_vals = [0.5 , 1 ,2 ,3 ,5 , 10 ]
for c_val in C_vals:
    svc = SVC(C = c_val ,kernel = "rbf")
    svc.fit(X_trained_scale , y_train )
    y_pred = svc.predict(X_test_scale)
    print ( "At C = ",c_val,"ACCURACY = ",accuracy_score(y_test , y_pred))
```
