# Stacking 
#### Hetrogenous ensemble learning

### Classifier 

```python
from sklearn.ensemble import StackingClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report
```

```python
X1 ,y1 = make_classification(
    n_samples = 500,
    n_features = 20,
    n_informative = 5,
    n_redundant  = 2,
    random_state = 42,
    
)
```

```python
X1_train , X1_test , y1_train , y1_test = train_test_split(
    X1,y1,test_size = 0.3,random_state = 42
)
```

```python
X1_train
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
```

```python
lr = LogisticRegression()
svc = SVC()
dtc = DecisionTreeClassifier(max_depth = 3)
```

```python
meta_model = LogisticRegression()# Default 

stacking_clf = StackingClassifier(
    estimators = [
        ("lr",lr),
        ("svc",svc),
        ("dtc",dtc)
    ],
    cv = 5, #Cross validation to prevent Overfitting 
    final_estimator = meta_model
)
```

```python
stacking_clf.fit(X1_train,y1_train)
```

```python
y1_pred = stacking_clf.predict(X1_test)

print("Accuracy Score : ",accuracy_score(y1_pred,y1_test))

print("CFR : ",classification_report(y1_pred,y1_test))
```

## Regression

```python
from sklearn.ensemble import StackingRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X2 ,y2 = make_regression(
    n_samples = 500,
    n_features = 20,
    n_informative = 5,
    random_state = 42,
    
)
```

```python
X2_train , X2_test , y2_train , y2_test = train_test_split(
    X2,y2,test_size = 0.3,random_state = 42
)
```

```python
X2_train
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
```

```python
lir = LinearRegression()
svr = SVR()
dtr = DecisionTreeRegressor(max_depth = 3)
```

```python
stacking_reg = StackingRegressor(
    estimators = [
        ("lir",lir),
        ("svr",svr),
        ("dtr",dtr)
    ],
    cv = 50, #Cross validation to prevent Overfitting 
)
```

```python
stacking_reg.fit(X2_train,y2_train)
```

```python
y2_pred = stacking_reg.predict(X2_test)
y2_pred_train = stacking_reg.predict(X2_train)

print("r2_score : ",r2_score(y2_test , y2_pred))
print("r2_score for train : ",r2_score(y2_train , y2_pred_train))
```
