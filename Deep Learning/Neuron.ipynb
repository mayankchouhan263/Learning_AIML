```python
import torch
import torch.nn as nn
```

```python
# INPUT 
torch.manual_seed(42) # Similar to random state 
inputs = torch.tensor([1.0, 2.0 ,3.0])
```

```python
neuron = nn.Linear(in_features = 3 , out_features = 1)
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# Verify Formula 

Actual_output= neuron.weight @ inputs + neuron.bias
```
