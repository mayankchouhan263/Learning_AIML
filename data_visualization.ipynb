```python
import matplotlib.pyplot as plt

# Create data
X = list(range(1, 101))
Y = [x**2 for x in X]
Z = [x**1.5 for x in X]
# Plot

plt.plot(X, Y)
plt.plot(X, Z)
plt.xlabel("X values")
plt.ylabel("Result")
plt.title("Plot of Y = X² and Z= X3")
plt.show()
```

```python
import numpy as np
X = [1,2,3,4,5,4]
Y = [2,5,4,8,9,6]
Z= [5,8,9,4,2,5]
width= 0.25
x= np.arange(len(Y))
plt.bar(x-width,Y,width, label ="Y")
plt.bar(x+width,X,width,label ="X")
plt.bar (x,Z,width, label ="Z")

plt.title("Bar Graph")
plt.xlabel("X values")
plt.ylabel("Result")
plt.legend()
plt.show()
```

```python
import numpy as np
np.random.seed(42)
data1 = np.random.normal(60, 10, 1000)
data2 = np.random.normal(70, 12, 1000)

plt.hist(data1, bins=30, density=True, alpha=0.6, label="Class A")
plt.hist(data2, bins=30, density=True, alpha=0.6, label="Class B")

mean1, std1 = np.mean(data1), np.std(data1)
mean2, std2 = np.mean(data2), np.std(data2)

plt.axvline(mean1, linestyle='--', linewidth=2, label=f"A Mean: {mean1:.1f}")
plt.axvline(mean2, linestyle='--', linewidth=2, label=f"B Mean: {mean2:.1f}")
plt.title("Advanced Histogram with Statistical Information")
plt.xlabel("Marks")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
direct = [50,60,70,80,90,100,110]
organic = [30,40,50,55,60,70,80]
social = [20,25,30,35,40,50,60]

x = np.arange(len(days))

# Subplot 1 - Line Graph
plt.subplot(2,2,1)
plt.plot(days, direct)
plt.title("Direct Traffic")

# Subplot 2 - Bar Chart
plt.subplot(2,2,2)
plt.bar(days, organic)
plt.title("Organic Traffic")

# Subplot 3 - Scatter Plot
plt.subplot(2,2,3)
plt.scatter(days, social)
plt.title("Social Traffic")

# Subplot 4 - Pie Chart (for Monday)
plt.subplot(2,2,4)
values = [direct[0], organic[0], social[0]]
plt.pie(values, labels=["Direct","Organic","Social"], autopct="%1.1f%%")
plt.title("Monday Traffic Distribution")

plt.tight_layout()
plt.show()
```
